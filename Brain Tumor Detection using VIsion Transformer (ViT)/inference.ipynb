```python
# Some metadata 

import torch
import sys
import platform
import transformers as t

print(f'------\n'
    f'Python version: {sys.version} \n'
      f'System: {platform.system()} \n'
      f'Platform: {platform.platform()}')

print(f'------\n'
      f'PyTorch version: {torch.__version__}\n'
      f'CUDA available: {torch.cuda.is_available()}\n'
      f'CUDA device count: {torch.cuda.device_count()}\n')

print(f'-------\n'
      f'Transformers version: {t.__version__}\n'
      f'-------\n')
```

```python
from transformers import ViTForImageClassification, ViTImageProcessor

model = ViTForImageClassification.from_pretrained("vit-braintumor-mri")
processor = ViTImageProcessor.from_pretrained("vit-braintumor-mri")
```

```python
import os
from PIL import Image
import torch

image_dir = "./test_images"
image_files = [f for f in os.listdir(image_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]

# Automatically fetch label names 
id2label = model.config.id2label  # Maps class index to label string

# Run inference
for image_file in image_files:
    image_path = os.path.join(image_dir, image_file)
    image = Image.open(image_path).convert("RGB")  

    # Preprocess
    inputs = processor(images=image, return_tensors="pt")

    # Inference
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits
        predicted_class_idx = logits.argmax(-1).item()
        predicted_label = id2label[predicted_class_idx]

    print(f"Image: {image_file} ==> Predicted class: {predicted_label.upper()}")
```

The model performs exceptionally well, as it manages to correctly label all the MRI scans. The true label can be seen in the image file name, "pi" for pituitary, and "gl" for glioma.
